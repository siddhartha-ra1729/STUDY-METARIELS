# How CG Works:--

# It’s a modification of the steepest descent algorithm.

# Instead of just using gradient directions, it uses conjugate directions to make optimization faster (especially for quadratic functions).

# CG requires the objective function to be differentiable.

```python
from scipy.optimize import minimize
def eqn(x):
    return x**2 + x + 2
mymin= minimize(eqn, 0, method='TNC')
print(mymin)
```

# TNC stands for Truncated Newton Conjugate-Gradient algorithm.

# It is used for minimizing scalar functions subject to bounds on variables, but no constraints.

```python
from scipy.optimize import minimize
def eqn(x):
    return x**2 + x + 2
mymin= minimize(eqn, 0, method='COBYLA')
print(mymin)                    
```

# COBYLA (Constrained Optimization BY Linear Approximations) is a gradient-free method (good for functions without derivatives).

# scipy cluster:

```python
import numpy as np
from scipy.cluster.vq import kmeans, whiten
import seaborn as sns
sns.set_style("darkgrid")

#create 50 datapoints in two clutster
pts=100
a=np.random.multivariate_normal([0,0], [[4,1],[1,4]], size=pts)
b=np.random.multivariate_normal([30,10], [[10,2],[2,1]], size=pts)
features= np.concatenate((a,b))

#whiten data
whitened= whiten(features)

#find 2 clusters in the data
codebook, distortion= kmeans(whitened,2)

#plt for whiten data and clusters centers in red
plt.scatter(whitened[:,0], whitened[:,1]) 
plt.scatter(codebook[:,0], codebook[:,1], c='r') 
plt.show()
```

# scipy.cluster.vq provides vector quantization tools: kmeans for clustering. & whiten for feature scaling.

# Whitening means scaling each feature to unit variance.

# This helps in clustering algorithms like KMeans, which are sensitive to feature scale.

# After this, all features contribute equally regardless of original units.



# codebook contains the centroids (cluster centers) → shape will be (2, 2)

# distortion is the sum of squared distances from each point to its closest center (a measure of how good the clustering is).

# Convex Hull: It is the smallest polygon that covers all the given point:

```python
from scipy.spatial import ConvexHull
import numpy as np
import matplotlib.pyplot as plt

new= np.array([[2,4],[3,4], [5,6], [1,5], [5,7],[4,5],[1,3],[9,8],[5,7],[1,4]])
hull= ConvexHull(new)
hull_points= hull.simplices
plt.scatter(new[:,0], new[:,1])
for i in hull_points:
    plt.plot(new[i,0], new[i,1], 'k-')
plt.show()    
```

# ConvexHull(new): It Calculates the smallest convex polygon (convex hull) that contains all the points.

# hull.simplices: Returns the vertex indices that form the edges of the convex hull.

# scipy-graph:

# whenever we working with graph we have module from scipy.
# In this we also have- an adjacency metrix: n*n matrix where n is the number of elements in the graph

```python
import numpy as np
from scipy. sparse.csgraph import connected_components
from scipy.sparse import csr_matrix 
new_value= np.array([[1,2,3],[1,0,0], [2,3,4]])
value_new= csr_matrix(new_value)
print(connected_components(value_new))
```

```python
import numpy as np
from scipy.sparse.csgraph import connected_components
from scipy.sparse import csr_matrix  # ✅ Correct import

# Sample adjacency matrix (should be square and represent a graph)
new_value = np.array([
    [0, 1, 1],
    [1, 0, 0],
    [1, 0, 0]
])

# Convert to sparse matrix
value_new = csr_matrix(new_value)

# Compute connected components
n_components, labels = connected_components(value_new)

print("Number of connected components:", n_components)
print("Component labels for each node:", labels)
```

# for dijkstra algo - we find the shortest path using this from one element to another  and for this its take 3 argu. limit, indices,  return_predecessors

```python
import numpy as np
from scipy. sparse.csgraph import dijkstra
from scipy.sparse import csr_matrix 
new_value= np.array([[1,2,3],[1,0,0], [2,3,4]])
value_new= csr_matrix(new_value)
print(dijkstra(value_new, return_predecessors=True,indices=0))
```

# floyd warshall-  In this also we use and we find the shortest path using  between all the pairs of elements 

```python
import numpy as np
from scipy. sparse.csgraph import floyd_warshall
from scipy.sparse import csr_matrix 
new_value= np.array([[1,2,3],[1,0,0], [2,3,4]])
value_new= csr_matrix(new_value)
print(floyd_warshall(value_new, return_predecessors=True))
```

# bellman ford-  In this also we use and we find the shortest path using  between all the pairs of elements 

```python
import numpy as np
from scipy. sparse.csgraph import bellman_ford
from scipy.sparse import csr_matrix 
new_value= np.array([[1,2,3],[1,0,0], [2,3,4]])
value_new= csr_matrix(new_value)
print(bellman_ford(value_new, return_predecessors=True, indices=0))
```
